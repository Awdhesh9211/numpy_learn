```python
# SPECIAL ARRAY
z=np.zeros((2,3))     # all 0
o=np.ones((2,2))      # all 1
e=np.eye(3)           # identity matrix
ar=np.arange(0, 10, 2) # [0,2,4,6,8]
lin=np.linspace(0, 1, 5) # 5 numbers between 0 and 1

print(z,o,e,ar,lin,sep="\n")
```

```python
# PROPERTIES
a=np.array([[5,4,3,2,4,2],[2,3,2,3,4,3],[2,2,3,2,3,2]])
print('shape= ',a.shape)    # dimensions (rows, cols)
print('dim= ',a.ndim)     # number of dimensions
print('size= ',a.size)     # total elements
print('dtype= ',a.dtype)    # data type
```

```python
# INDEXING
a=np.array([2,2,3,2,3,2])
print(a[0])   
print(a[a!=3])        
```

```python
# SLICING
a=np.arange(1,31).reshape(6,5)
print(a)
print(a[0:2,1:2])  #perticular sub matrix
print(a[:,2].reshape(6,1))   #column
print(a[2,:])  #row
```

```python
# RESHAPE
a = np.arange(6)

a.reshape(2,3)
```

```python
#  OPERATION
a = np.array([1,2,3])
b = np.array([4,5,6])

a + b     # [5,7,9]
a * b     # [4,10,18]
a ** 2    # [1,4,9]
```

```python
#  STATISTICS
print(np.sum(a))
print(np.mean(a))
print(np.min(a))
print(np.max(a))
print(np.std(a))
```

```python
#  BROADCASTING 
a = np.array([1,2,3])
b = 2

a + b   # [3,4,5]   WHEN operation [1,2,3] + [2,2,2] (THIS IS BROADCASTING )
a / b
a // b
a * b
a ** b
```

```python
# ITERATION 
for x in a:
    print(x)
```

```python
#  LINEAR ALGEBRA 
a = np.array([[1,2],[3,4]])
b = np.array([[5,6],[7,8]])

print(np.dot(a, b)      )  # matrix multiplication
print(np.linalg.inv(a)   ) # inverse
print(np.linalg.det(a)    )# determinant
```

```python

#  CRUD
arr = np.array([1,2,3,4,5])

# CREATE
arr = np.append(arr, 6)

# READ
print(arr[arr % 2 == 0])

# UPDATE
arr[arr > 3] = 0

# DELETE
arr = np.delete(arr, 1)
```

```python

# =========================
# CREATE MATRICES
# =========================
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

print("Matrix A:\n", A)
print("Matrix B:\n", B)

# =========================
# ADDITION & SUBTRACTION
# =========================
print("\nA + B:\n", A + B)
print("A - B:\n", A - B)

# =========================
# ELEMENT-WISE MULTIPLICATION
# =========================
print("\nElement-wise A * B:\n", A * B)

# =========================
# MATRIX MULTIPLICATION
# =========================
print("\nMatrix Multiplication (A @ B):\n", A @ B)
print("Using dot():\n", np.dot(A, B))

# =========================
# TRANSPOSE
# =========================
print("\nTranspose of A:\n", A.T)

# =========================
# DETERMINANT
# =========================
print("\nDeterminant of A:", np.linalg.det(A))

# =========================
# INVERSE
# =========================
print("\nInverse of A:\n", np.linalg.inv(A))

# =========================
# IDENTITY MATRIX
# =========================
I = np.eye(2)
print("\nIdentity Matrix:\n", I)

# =========================
# MATRIX POWER
# =========================
print("\nA squared (A^2):\n", np.linalg.matrix_power(A, 2))

# =========================
# TRACE (sum of diagonal)
# =========================
print("\nTrace of A:", np.trace(A))

# =========================
# EIGEN VALUES & VECTORS
# =========================
eigenvalues, eigenvectors = np.linalg.eig(A)
print("\nEigenvalues:\n", eigenvalues)
print("Eigenvectors:\n", eigenvectors)

# =========================
# RANK
# =========================
print("\nRank of A:", np.linalg.matrix_rank(A))

# =========================
# SOLVE LINEAR EQUATION
# Ax = b
# =========================
b = np.array([5, 11])
x = np.linalg.solve(A, b)
print("\nSolution of Ax = b:\n", x)

# =========================
# STACKING MATRICES
# =========================
print("\nVertical Stack:\n", np.vstack((A, B)))
print("\nHorizontal Stack:\n", np.hstack((A, B)))

# =========================
# RESHAPE
# =========================
C = np.arange(1, 7)
print("\nReshape 1D to 2x3:\n", C.reshape(2,3))

# =========================
# BROADCASTING
# =========================
D = np.array([1, 2])
print("\nBroadcasting A + D:\n", A + D)

# =========================
# MAX, MIN, SUM (AXIS)
# =========================
print("\nColumn-wise sum:\n", np.sum(A, axis=0))
print("Row-wise sum:\n", np.sum(A, axis=1))
print("Max element:", np.max(A))
print("Min element:", np.min(A))
```
